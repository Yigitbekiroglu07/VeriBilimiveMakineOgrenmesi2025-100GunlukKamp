```python
df.drop("No", axis = 1, inplace= True)
```

```python
df.rename(columns={
    "X1 transaction date": "Transaction Date",
    "X2 house age": "House Age",
    "X3 distance to the nearest MRT station": "Distance to the Nearest MRT Station",
    "X4 number of convenience stores": "Number of Convenience Stores",
    "X5 latitude": "Latitude",
    "X6 longitude": "Longitude",
    "Y house price of unit area": "House Price of Unit Area"
}, inplace=True)
```

```python
df.head()
```

```python
# dependent and independent features
X = df.drop("House Price of Unit Area", axis=1)
y = df["House Price of Unit Area"]
```

```python
# train - test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size= 0.25, random_state=15)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
X_train_scaled
```

```python
X_test_scaled
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
tree_model = DecisionTreeRegressor(criterion= "squared_error", splitter= "random", max_depth=None, min_samples_split = 2, min_samples_leaf=1,random_state=None)
tree_model.fit(X_train_scaled, y_train)
y_pred = tree_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
print(r2_score(y_test,y_pred))
```

```python
## Hyperparameter Tuning using GridSearch CV
from sklearn.model_selection import GridSearchCV
 
# defining parameter range
param_grid = {'criterion': ['squared_error','riedman_mse','absolute_error','poisson'],
              'splitter': ['best','random'],
              'max_depth': [0.3,3,30,300,None],
             'min_samples_split': [0.2,2,20,200],
            'min_samples_leaf': [0.1,1,10,100],
              'max_features': ['sqrt','log2',None],
             'random_state': [0.1,10,100,1000,None]}
```

```python
grid = GridSearchCV(DecisionTreeRegressor(), param_grid, refit = True,verbose = 3,n_jobs=-1) # n_jobs=-1 means use all processor
```

```python
grid.fit(X_train_scaled, y_train)
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(X_test_scaled)
```

```python
print(mean_absolute_error(y_test,y_pred)) 
print(mean_squared_error(y_test,y_pred)) 
print(r2_score(y_test,y_pred))
```
